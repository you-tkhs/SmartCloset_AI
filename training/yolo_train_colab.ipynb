```python
from ultralytics import YOLO
from google.colab import runtime
```

- 学習実行(yolov8n-segを用いて実行)

```python
model = YOLO("yolov8n-seg.pt")

results = model.train(
    data="/content/dataset/fashionpedia_yolo_9class/fashion.yaml",
    epochs=1,
    imgsz=640,
    batch=32,
    device=0,
    workers=2,

    # Drive保存（超重要）
    project="/content/drive/MyDrive/コーディング/web開発/smartcloset_AI/finetuning/",
    name="fashionpedia_seg_v4"
)

runtime.unassign()
```

・学習実行(yolo8n-segを用いてデータ拡張も加味した実行)

```python
model = YOLO("yolov8n-seg.pt")

results = model.train(
    data="/content/dataset/fashionpedia_yolo_9class/fashion.yaml",
    epochs=30,
    imgsz=640,
    batch=64,
    patience=10,
    workers=2,
    device=0,

    # augmentation
    hsv_h=0.015,
    hsv_s=0.6,
    hsv_v=0.4,
    degrees=8.0,
    translate=0.15,
    scale=0.6,
    shear=2.0,
    perspective=0.0005,
    fliplr=0.5,
    flipud=0.0,

    mosaic=1.0,
    close_mosaic=5,
    copy_paste=0.1,

    project="/content/drive/MyDrive/コーディング/web開発/smartcloset_AI/finetuning/",
    name="fashionpedia_seg_v4"
)
```

- 学習実行(学習が終了したモデルの初期値を用いて実行)

```python
model = YOLO(
"/content/drive/MyDrive/コーディング/web開発/smartcloset_AI/finetuning/fashionpedia_seg_v2/weights/best.pt"
)

results = model.train(
    data="/content/dataset/fashionpedia_yolo/fashion.yaml",
    epochs=20,
    imgsz=640,
    batch=32,
    patience=5,
    workers=2,
    device=0,

    project="/content/drive/MyDrive/コーディング/web開発/smartcloset_AI/finetuning",
    name="fashionpedia_seg_v4_tmp"
)

runtime.unassign()
```

- 学習実行(ランタイム切れによる学習途中の重みからの実行)

```python
model = YOLO(
"/content/drive/MyDrive/コーディング/web開発/smartcloset_AI/finetuning/fashionpedia_seg_v4-2/weights/last.pt"
)

model.train(resume=True)

runtime.unassign()
```
